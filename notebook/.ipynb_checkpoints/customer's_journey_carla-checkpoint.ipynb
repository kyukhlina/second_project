#### How should customer journeys be measured?

In this analysis, a customer journey represents the sequence of interactions performed by a customer while attempting to complete the online process. We initially consider each `visit_id` as a candidate customer journey, with its events ordered chronologically using `date_time`.

```python
import pandas as pd

url_web_1 = "https://raw.githubusercontent.com/data-bootcamp-v4/lessons/refs/heads/main/5_6_eda_inf_stats_tableau/project/files_for_project/df_final_web_data_pt_1.txt"
url_web_2 = "https://raw.githubusercontent.com/data-bootcamp-v4/lessons/refs/heads/main/5_6_eda_inf_stats_tableau/project/files_for_project/df_final_web_data_pt_2.txt"

web_1 = pd.read_csv(url_web_1, sep=",")
web_2 = pd.read_csv(url_web_2, sep=",")

df_final_web_data = pd.concat([web_1, web_2], ignore_index=True)
df_final_web_data.shape
```

```python
df_clean_web_data = df_final_web_data.drop_duplicates().reset_index(drop=True)
```

```python
df_clean_web_data = df_clean_web_data.sort_values(
    ["client_id", "visitor_id", "visit_id", "date_time"]).reset_index(drop=True)
df_clean_web_data
```
