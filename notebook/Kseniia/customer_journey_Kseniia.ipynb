## Adding filters for process steps

```python
df_all = df_all.sort_values(['client_id', 'visit_id', 'date_time'])

#The previos step for each session
df_all['prev_step'] = (df_all.groupby('visit_id')['process_step'].shift(1)).fillna(0).map(int)
```

```python
df_all.head(3)
```

```python
#Difference of step numbers to identify backward steps. If 'df_all['diff'] < 0' there was a step for 1 ore more position back
df_all['diff'] = df_all['process_step'] - df_all['prev_step']
df_all[df_all['diff'] == -1].head(3)
```

```python
df_all['back_more_than_1_step'] = (df_all['diff'] < -1) & (~df_all['prev_step'])
df_all[df_all['back_more_than_1_step'] == True].head(3)
```

```python
df_all['date_time'] = pd.to_datetime(df_all['date_time'])
```

```python
df_all = df_all.sort_values(by=['visit_id', 'date_time']).reset_index(drop=True)
df_all['time_to_next'] = (df_all.groupby('visit_id')['date_time'].diff().dt.total_seconds()).fillna(0).map(int)
df_all['is_backward'] = df_all['diff'] < 0
```

```python
client_journey = df_all.groupby('visit_id').agg(journey=('process_step', list), 
        session_sec=('time_to_next', 'sum'),
        date_time_start=('date_time', 'min'),
        date_time_stop=('date_time', 'max'),
        last_step=('process_step', 'last'),
        first_step=('process_step', 'first'),
        has_backward_nav=('is_backward', 'any'),
        variation=('variation', 'first')).reset_index()
```

```python
client_journey['is_ideal'] = client_journey['journey'].apply(lambda x: x == [1,2,3,4,5])
client_journey['contains_5'] = client_journey['journey'].apply(lambda x: {5}.issubset(x))
client_journey.head(3)
```
