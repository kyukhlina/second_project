Which identifier should be used for each type of analysis?

```python
# Check for unique values for column client, visitor and visit

print("Unique client_id:", df2["client_id"].nunique())
print("Unique visitor_id:", df2["visitor_id"].nunique())
print("Unique visit_id:", df2["visit_id"].nunique())
print("Rows:", len(df2))
```

One client has many visit --> Unique client_id: 67,430 != Unique visit_id: 82,841

```python
# Check if one client has multiple visits

df2.groupby("client_id")["visit_id"].nunique().sort_values(ascending=False).head()
```

Customer journey --> A client can return multiple times and create multiple visits.

```python
# Client --> Visitor

df2.groupby("client_id")["visitor_id"].nunique().value_counts().sort_index()
```

Verifying Hierarchy 

```python
# Visit --> Client 

df2.groupby("visit_id")["client_id"].nunique().value_counts()
```

A visit should ideally belong to only one client.

```python
# Visit --> Visitor 

df2.groupby("visit_id")["visitor_id"].nunique().value_counts()
```

One visit id per visitor

```python
# one visitor belongs to multiple clients

df2.groupby("visitor_id")["client_id"].nunique().value_counts()
```

```python
Visitor_id are associated with multiple clients therefore should be avoided as a primary customer identifier (could be due to shared devices, authentication, household accounts) 
```

```python
# Checking for exact duplicates

df2[df2.duplicated(keep=False)].sort_values(
["client_id", "visit_id"]
)
```

```python
# Check repeated process steps within a visit

duplicates_steps = df2[
    df2.duplicated(
        subset=["client_id", "visit_id", "process_step"],
        keep=False
)
]

duplicates_steps.sort_values(
["client_id", "visit_id", "process_step", "date_time"]
)
```

```python
# How often it happens
(
df2.groupby(["client_id", "visit_id", "process_step"])
    .size()
    .sort_values(ascending=False)
    .head(20)
)
```

```python
# inspect time differences 

# convert time

df2["date_time"] = pd.to_datetime(df2["date_time"])
```

```python
sample = duplicates_steps.sort_values(
    ["client_id", "visit_id", "process_step", "date_time"]
)

sample.head(10)
```
