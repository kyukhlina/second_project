### Vanguard Project - Client Profiles df

This notebook focuses on the analysis of the df_final_demo dataset, which contains demographic and account information about Vanguard clients.

The main objectives are to understand the structure and content of the dataset, assess its data quality, identify potential inconsistencies, and determine whether any issues could affect the subsequent analysis.

```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/data-bootcamp-v4/lessons/refs/heads/main/5_6_eda_inf_stats_tableau/project/files_for_project/df_final_demo.txt"
df_final_demo = pd.read_csv(url, sep=",")
df_final_demo.head()
```

```python
df_final_demo.info()
```

Most variables are numerical, while `gendr` is categorical. The `client_id` variable contains no missing values, whereas the remaining variables contain a small number of missing observations.

```python
df_final_demo.isna().sum()
```

```python
df_final_demo[df_final_demo.isna().any(axis=1)]
```

We identified 15 clients with missing information. For 14 clients, all profile variables are missing, while one client has only the age missing.

```python
df_final_demo["client_id"].nunique() == df_final_demo.shape[0]
```

```python
df_final_demo[["clnt_tenure_yr", "clnt_tenure_mnth"]].describe()
```

```python
df_final_demo["tenure_diff"] = (df_final_demo["clnt_tenure_mnth"] - df_final_demo["clnt_tenure_yr"] * 12)
df_final_demo["tenure_diff"].describe()
```

```python
df_final_demo[
    ["clnt_tenure_yr", "clnt_tenure_mnth", "tenure_diff"]
].head(20)
```

The difference between total months and years converted to months ranges from 0 to 12 months.

This suggests that `clnt_tenure_yr` represents completed years of tenure, while `clnt_tenure_mnth` represents total tenure in months. 

Therefore, the apparent difference between the two variables does not indicate a data inconsistency. The tenure variables appear to be consistent.

```python
df_final_demo.drop(columns="tenure_diff", inplace=True)
```

```python
profile_columns = ["clnt_tenure_yr", "clnt_tenure_mnth", "clnt_age", "gendr", "num_accts", "bal", "calls_6_mnth", "logons_6_mnth"]
missing_profile = df_final_demo[df_final_demo[profile_columns].isna().all(axis=1)]
missing_profile
```

```python
missing_profile["client_id"]
```

#### Investigation missing profile information

```python
url_2 = "https://raw.githubusercontent.com/data-bootcamp-v4/lessons/refs/heads/main/5_6_eda_inf_stats_tableau/project/files_for_project/df_final_experiment_clients.txt"
df_final_experiment_clients = pd.read_csv(url_2, sep=",")
df_final_experiment_clients.head()
```

```python
df_final_experiment_clients.shape
```

```python
missing_profile_experiment = df_final_experiment_clients[
    df_final_experiment_clients["client_id"].isin(missing_profile["client_id"])
]

missing_profile_experiment
```

All 14 clients with completely missing profile information are present in the experiment roster.

Among these clients, 6 are assigned to the Control group, 6 to the Test group, and 2 have no experimental variation recorded.

```python
df_final_experiment_clients["Variation"].value_counts(dropna=False)
```

The `client_id` is consistent across the two datasets: all 70,609 clients in `df_final_demo` are also present in `df_final_experiment_clients`.

However, the experimental assignment is not complete. While 26,968 clients are assigned to the Test group and 23,532 to the Control group, 20,109 clients have no recorded value for `variation`.

```python
clients_no_variation = df_final_experiment_clients[df_final_experiment_clients["Variation"].isna()]
clients_no_variation
```

```python
demo_no_variation = df_final_demo[df_final_demo['client_id'].isin(clients_no_variation['client_id'])]
demo_no_variation.describe()
```
