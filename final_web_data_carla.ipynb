### df_final_web_data

```python
import pandas as pd

url_web_1 = "https://raw.githubusercontent.com/data-bootcamp-v4/lessons/refs/heads/main/5_6_eda_inf_stats_tableau/project/files_for_project/df_final_web_data_pt_1.txt"
url_web_2 = "https://raw.githubusercontent.com/data-bootcamp-v4/lessons/refs/heads/main/5_6_eda_inf_stats_tableau/project/files_for_project/df_final_web_data_pt_2.txt"

web_1 = pd.read_csv(url_web_1, sep=",")
web_2 = pd.read_csv(url_web_2, sep=",")

df_final_web_data = pd.concat([web_1, web_2], ignore_index=True)
df_final_web_data.shape
```

```python
df_clean_web_data = df_final_web_data.drop_duplicates().reset_index(drop=True)
```

```python
print(df_final_web_data.shape)
print(df_clean_web_data.shape)
```

We checked for exact duplicate rows, considering all columns (`client_id`, `visitor_id`, `visit_id`, `process_step` and `date_time`).

A total of 10.764 exact duplicate events were removed because they represent identical event records and could otherwise artificially increase the number of interactions.

Convert date_time column to a datedime type

```python
df_clean_web_data["date_time"]=pd.to_datetime(df_clean_web_data["date_time"]).reset_index(drop=True)
df_clean_web_data["date_time"].dtype
```

```python
df_clean_web_data.sort_values(["client_id", "visitor_id", "visit_id", "date_time"]).reset_index(drop=True)
```

```python
client_summary=(df_clean_web_data.groupby("client_id").agg(num_visitors=("visitor_id","nunique"), num_visits=("visit_id","nunique")).reset_index())
client_summary.sample(20)
```
